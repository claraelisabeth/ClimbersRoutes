```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
df_climbers = pd.read_csv("data/8a/climber_df.csv")
df_grades_conversion = pd.read_csv("data/8a/grades_conversion_table.csv")
df_routes = pd.read_csv("data/8a/routes_rated.csv")
```

```python
display(df_climbers.info())
display(df_climbers.head(5))
```

```python
display(df_routes.info())
display(df_routes.head(5))
```

```python
display(df_grades_conversion.info())
display(df_grades_conversion.head(5))
```

```python
df_climbers = df_climbers.drop(['years_cl', 'date_first', 'date_last'], axis=1)
df_climbers['age'] = df_climbers['age'].astype(int)

df_routes = df_routes.drop(['Unnamed: 0', 'cluster', 'tall_recommend_sum', 'rating_tot'], axis=1)
df_routes['grade_mean'] = [round(grade) for grade in df_routes['grade_mean']]
df_routes['grade_mean'] = df_routes['grade_mean'].astype(int)
df_routes.rename(columns={
    'grade_mean': 'grade_id',
    'name' : 'route'},
    inplace=True)
df_routes['country'] = df_routes['country'].str.upper()

df_grades_conversion = df_grades_conversion.drop(['Unnamed: 0'], axis=1)
df_grades_conversion = df_grades_conversion[df_grades_conversion['grade_id']>=3]
```

```python
print("Number of unique climbers: ", len(df_climbers['user_id'].unique()))
print("Number of unique routes: ", len(df_routes['name_id'].unique()))
print("Number of unique grades: ", len(df_grades_conversion['grade_fra'].unique()))
print("Number of route countries: ", len(df_routes['country'].unique()))
print("Number of climbers home countries: ", len(df_climbers['country'].unique()))
```

```python
display(len(df_climbers[df_climbers['grades_count']==0]))
display(len(df_climbers[df_climbers['age']<10]))
display(len(df_climbers[df_climbers['weight']<40]))
display(len(df_climbers[df_climbers['height']<120]))
```

```python
df_routes = pd.merge(df_routes, df_grades_conversion, on = 'grade_id')
df_routes = df_routes.drop(['grade_id'], axis = 1)
```

```python
df_climbing_history = pd.read_csv("climbing_history_clean.csv")
```

```python
display(df_climbing_history.info())
display(df_climbing_history.head(5))
```

```python
df_climbing_history['route'] = df_climbing_history['route'].str.lower()
```

```python
df_climbing_history_merge = pd.merge(df_routes, df_climbing_history, on = 'route')
df_climbing_history_merge = df_climbing_history_merge.drop(['link', 'parsed_dates', 'date', 'name_id', 'official_grade', 'ascent_date'], axis = 1)
```
